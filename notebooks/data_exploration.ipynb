```python
# Importing necessary libraries
import nfl_data_py as nfl
import pandas as pd
```

```python
# Set display precision for readability
pd.set_option("display.precision", 1)
YEAR = [2024]
# Inital look of the season-level stats
nfl.import_seasonal_data(YEAR).head()
```

```python
# Pulling weekly stats for the 2024 season
weekly = nfl.import_weekly_data(YEAR)
print(weekly.head())
```

```python
# Rows and Columns
print(weekly.shape)
```

```python
# List of every Column
print(weekly.columns)
```

```python
# Every unique player name
print(weekly["player_display_name"].unique())
```

```python
# Number of distinct players
print(weekly["player_display_name"].unique().shape)
```

```python
# Position labels
print(weekly["position"].unique())
```

```python
# Position groups
print(weekly["position_group"].unique())
```

```python
# Count rows per position
print(weekly['position'].value_counts())
```

```python
# Narrow down to only the columns we actually need going forward, drop the rest
weekly = weekly[["player_display_name",
"position",
"recent_team",
"season_type",
"fantasy_points_ppr"]]

print(weekly)
```

```python
# Season types
print(weekly["season_type"].unique())
```

```python
# Drop playoff stats and keep regual season stats
weekly = weekly[weekly["season_type"] == "REG"]
print(weekly)
```

```python
# Keep only the four offensive skill positions
keep_positions = ["WR", "RB", "TE", "QB"]
weekly = weekly[weekly["position"].isin(keep_positions)]
print(weekly)
```

```python
# Sum each player's weekly points into a season total, sorted highest to lowest
weekly = weekly.groupby(
    ["player_display_name", "position", "recent_team"],
    as_index=False
)["fantasy_points_ppr"].sum()
weekly = weekly.sort_values(by="fantasy_points_ppr", ascending=False)
weekly = weekly.reset_index(drop=True)
print(weekly)
```

```python
# Check for players appearing more than once -- flags players traded mid-season
name_counts = weekly['player_display_name'].value_counts()
name_counts[name_counts > 1]
```

```python
# Fix the trade-splitting problem
# Sum points by name + position only
points_totals = weekly.groupby(
    ['player_display_name', 'position'],
    as_index=False
)['fantasy_points_ppr'].sum()
# Compute each player's most-played team
primary_team = (
    weekly.groupby(['player_display_name', 'recent_team'])
    .size()
    .reset_index(name='weeks_with_team')
    .sort_values('weeks_with_team', ascending=False)
    .drop_duplicates(subset='player_display_name', keep='first')
    [['player_display_name', 'recent_team']]
)
# Merge the two together
final = points_totals.merge(primary_team, on='player_display_name', how='left')
final = final.sort_values('fantasy_points_ppr', ascending=False).reset_index(drop=True)
final
```

```python
#Check for duplicates
final['player_display_name'].value_counts().loc[lambda x: x > 1]
```

```python
# Original weekly data
raw = nfl.import_weekly_data(YEAR)
raw = raw[raw["season_type"] == "REG"]
raw = raw[raw["position"].isin(["WR", "RB", "TE", "QB"])]
# Count how many distinct weeks each player played
games_played = (
    raw.groupby("player_display_name")["week"]
    .nunique()
    .reset_index(name="games_played")
)
games_played.head()
```

```python
# Attach the games_played count onto each player's row in `final`
final = final.merge(games_played, on="player_display_name", how="left")
final
```

```python
# Look at the distribution of games_played to decide where to draw the minimum-games cutoff
final["games_played"].describe()
final["games_played"].value_counts().sort_index()
```

```python
# Drop anyone who played fewer than 4 games (filters out small-sample noise)
final = final[final["games_played"] >= 4].reset_index(drop=True)
final.shape
```

```python
# Confirm how many players remain at each position after filtering
final["position"].value_counts()
```

```python
# Build a team x position grid to check whether every team still has depth at every position
team_position_counts = final.groupby(["recent_team", "position"]).size().unstack(fill_value=0)
team_position_counts[(team_position_counts == 0).any(axis=1)]
team_position_counts
```
